```python
# ============
# CODE_BLOCK_D
# ============

# =========================================
# MARKDOWN CELL TEMPLATE - CHAPTER 4 TABLES
# =========================================

MARKDOWN_TEMPLATE_D = """## NB07 - Chapter 4 Tables

This block loads, formats, and displays Table 16 through Table 25 from saved CSV artifacts only.
"""

display(Markdown(MARKDOWN_TEMPLATE_D))

# ===============================================================
# DISPLAY THE GLOBAL NOTES ABOUT WINDOW LENGTHS AND ENTROPY TIERS
# ===============================================================

GLOBAL_NOTES_D = """
**Important window note:** Forecast evaluation tables use a **500-observation** forecast window. Portfolio tables use a **499-trading-day** active backtest window.

**Entropy tier note:** **Tier 1** = feature-importance entropy from **NB04**. **Tier 2** = factor-exposure entropy from **NB06**.
"""

display(Markdown(GLOBAL_NOTES_D))

# ====================================================
# DEFINE THE TABLE MAP FOR REPORT TABLES 16 THROUGH 25
# ====================================================

TABLE_MAP_D = [
    (16, "Causal versus baseline RMSE and MAE by ticker and aggregate", TABLES_DIR / "notebook04_causal_vs_baseline_rmse_mae.csv"),
    (17, "Asset-class decomposition of RMSE and MAE deltas", TABLES_DIR / "notebook04_asset_class_comparison.csv"),
    (18, "Constrained feature importance for SPY, TLT, and GLD", TABLES_DIR / "notebook04_causal_feature_importance.csv"),
    (19, "Tier 1 entropy comparison - constrained versus unconstrained", TABLES_DIR / "notebook04_entropy_comparison.csv"),
    (20, "DAG gating summary - allowed versus excluded features", TABLES_DIR / "notebook04_dag_gating_summary.csv"),
    (21, "Portfolio performance for all seven model variants", TABLES_DIR / "notebook05_portfolio_performance.csv"),
    (22, "Risk-targeting stability for all seven model variants", TABLES_DIR / "notebook05_risk_targeting_stability.csv"),
    (23, "Turnover summary for all seven model variants", TABLES_DIR / "notebook05_turnover_summary.csv"),
    (24, "Ablation summary across executed stages", TABLES_DIR / "notebook06_ablation_table.csv"),
    (25, "Diebold-Mariano HAC test results", TABLES_DIR / "notebook06_dm_test_results.csv"),
]

for _table_number, _table_title, _table_path in TABLE_MAP_D:
    if not _table_path.exists():
        raise FileNotFoundError(f"MISSING REQUIRED TABLE FOR TABLE {_table_number}: {_table_path}")

# ==================================================================
# DEFINE A HELPER THAT CONVERTS RAW COLUMN NAMES INTO DISPLAY LABELS
# ==================================================================

def prettify_column_name(column_name):

    # ==========================================================
    # STANDARDIZE THE RAW COLUMN NAME FOR HUMAN-READABLE DISPLAY
    # ==========================================================

    label = str(column_name).replace("__", " ").replace("_", " ").strip().title()

    # ====================================================
    # REPLACE COMMON METRIC TOKENS WITH UPPERCASE ACRONYMS
    # ====================================================

    replacements = {
        "Rmse": "RMSE",
        "Mae": "MAE",
        "Dm": "DM",
        "Hac": "HAC",
        "Vix": "VIX",
        "Ewma": "EWMA",
        "Xgboost": "XGBOOST",
        "Mlp": "MLP",
        "Lstm": "LSTM",
        "Spy": "SPY",
        "Tlt": "TLT",
        "Gld": "GLD",
        "Hyg": "HYG",
        "Lqd": "LQD",
        "Qqq": "QQQ",
        "Iwm": "IWM",
        "Efa": "EFA",
        "Eem": "EEM",
        "Xlk": "XLK",
        "Xlf": "XLF",
        "Xle": "XLE",
        "Xlv": "XLV",
        "Dbc": "DBC",
        "Hml": "HML",
        "Smb": "SMB",
        "Rf": "RF",
    }

    for _old_text, _new_text in replacements.items():
        label = label.replace(_old_text, _new_text)

    return label

# ========================================================
# DEFINE A HELPER THAT LOADS, CLEANS, AND ROUNDS ONE TABLE
# ========================================================

def load_and_format_table(csv_path):

    # ============================
    # LOAD ONE CSV TABLE FROM DISK
    # ============================

    df = pd.read_csv(csv_path)

    # =================================================
    # ROUND EVERY NUMERIC COLUMN TO FOUR DECIMAL PLACES
    # =================================================

    numeric_columns = df.select_dtypes(include="number").columns.tolist()
    if numeric_columns:
        df[numeric_columns] = df[numeric_columns].round(4)

    # ===================================================
    # APPLY HUMAN-READABLE DISPLAY LABELS TO EVERY COLUMN
    # ===================================================

    df = df.rename(columns={_column: prettify_column_name(_column) for _column in df.columns})

    return df

# =============================================================
# DEFINE A HELPER THAT DISPLAYS ONE REPORT TABLE WITH A HEADING
# =============================================================

def display_report_table(table_number, table_title, formatted_df):

    # ============================================================
    # DISPLAY THE REPORT TABLE HEADING AND THE FORMATTED DATAFRAME
    # ============================================================

    display(Markdown(f"### TABLE {table_number} - {table_title}"))
    display(formatted_df)

# ==================================================
# LOAD, FORMAT, AND DISPLAY ALL TEN CHAPTER 4 TABLES
# ==================================================

FORMATTED_TABLES_D = {}
RAW_TABLES_D = {}

for _table_number, _table_title, _table_path in TABLE_MAP_D:
    RAW_TABLES_D[_table_number] = pd.read_csv(_table_path)
    FORMATTED_TABLES_D[_table_number] = load_and_format_table(_table_path)
    display_report_table(
        table_number=_table_number,
        table_title=_table_title,
        formatted_df=FORMATTED_TABLES_D[_table_number],
    )

# =================================================================
# BUILD THE CONDENSED NB07 FINAL SUMMARY FROM TABLE 21 AND TABLE 16
# =================================================================

TABLE21_RAW_DF = RAW_TABLES_D[21].copy()
TABLE16_RAW_DF = RAW_TABLES_D[16].copy()

# ===========================================================
# ATTEMPT TO EXTRACT AGGREGATE FORECAST METRICS FROM TABLE 16
# ===========================================================

def extract_forecast_summary(table16_df):

    # ============================================================
    # IDENTIFY THE MOST LIKELY MODEL COLUMN IN THE INPUT DATAFRAME
    # ============================================================

    model_column = next((
        _column
        for _column in table16_df.columns
        if "model" in _column.lower() or "variant" in _column.lower()
    ), None)

    # =====================================================
    # IDENTIFY THE MOST LIKELY TICKER OR AGGREGATION COLUMN
    # =====================================================

    aggregate_column = next((
        _column
        for _column in table16_df.columns
        if "ticker" in _column.lower() or "asset" in _column.lower() or "group" in _column.lower()
    ), None)

    # =============================================
    # IDENTIFY THE MOST LIKELY RMSE AND MAE COLUMNS
    # =============================================

    rmse_column = next((_column for _column in table16_df.columns if "rmse" in _column.lower()), None)
    mae_column = next((_column for _column in table16_df.columns if "mae" in _column.lower()), None)

    # ===========================================================
    # EXTRACT THE AGGREGATE ROW WHEN AN AGGREGATION COLUMN EXISTS
    # ===========================================================

    working_df = table16_df.copy()
    if aggregate_column is not None:
        aggregate_mask = working_df[aggregate_column].astype(str).str.upper().isin([
            "AGGREGATE", "ALL", "OVERALL", "MEAN", "AVG", "AVERAGE"
        ])
        if aggregate_mask.any():
            working_df = working_df.loc[aggregate_mask].copy()

    # ==========================================================
    # RETURN A NARROW MODEL LEVEL FORECAST SUMMARY WHEN POSSIBLE
    # ==========================================================

    if model_column is not None and rmse_column is not None and mae_column is not None:
        narrow_df = working_df[[model_column, rmse_column, mae_column]].copy()
        narrow_df = narrow_df.rename(columns={
            model_column: "MODEL",
            rmse_column: "FORECAST_RMSE_500_OBS",
            mae_column: "FORECAST_MAE_500_OBS",
        })
        narrow_df["MODEL"] = narrow_df["MODEL"].astype(str).str.upper()
        narrow_df = narrow_df.drop_duplicates(subset=["MODEL"])
        return narrow_df

    # ====================================================================
    # SYNTHESIZE A WIDE FORMAT FORECAST SUMMARY WHEN THE TABLE IS NOT TIDY
    # ====================================================================

    if not working_df.empty:
        first_row = working_df.iloc[0]
        synthesized_rows = []

        model_keyword_map = {
            "EWMA": ["ewma"],
            "XGBOOST": ["xgboost", "xgb"],
            "CAUSAL_XGBOOST": ["causal_xgboost", "causal", "xgboost"],
        }

        for _model_name, _keywords in model_keyword_map.items():
            rmse_value = pd.NA
            mae_value = pd.NA

            for _column in working_df.columns:
                column_lower = _column.lower()
                if "rmse" in column_lower and any(_keyword in column_lower for _keyword in _keywords):
                    rmse_value = first_row[_column]
                if "mae" in column_lower and any(_keyword in column_lower for _keyword in _keywords):
                    mae_value = first_row[_column]

            synthesized_rows.append({
                "MODEL": _model_name,
                "FORECAST_RMSE_500_OBS": rmse_value,
                "FORECAST_MAE_500_OBS": mae_value,
            })

        return pd.DataFrame(synthesized_rows)

    return pd.DataFrame(columns=["MODEL", "FORECAST_RMSE_500_OBS", "FORECAST_MAE_500_OBS"])

FORECAST_SUMMARY_DF = extract_forecast_summary(TABLE16_RAW_DF)

# =================================================================
# BUILD THE FINAL SUMMARY BASE FROM THE PORTFOLIO PERFORMANCE TABLE
# =================================================================

model_column_table21 = next((
    _column
    for _column in TABLE21_RAW_DF.columns
    if "model" in _column.lower() or "variant" in _column.lower()
), TABLE21_RAW_DF.columns[0] if len(TABLE21_RAW_DF.columns) > 0 else "MODEL")

NB07_SUMMARY_DF = TABLE21_RAW_DF.copy()
NB07_SUMMARY_DF = NB07_SUMMARY_DF.rename(columns={model_column_table21: "MODEL"})
NB07_SUMMARY_DF["MODEL"] = NB07_SUMMARY_DF["MODEL"].astype(str).str.upper()

# ================================================================
# MERGE FORECAST METRICS INTO THE PORTFOLIO SUMMARY WHEN AVAILABLE
# ================================================================

NB07_SUMMARY_DF = NB07_SUMMARY_DF.merge(
    FORECAST_SUMMARY_DF,
    on="MODEL",
    how="left",
)

numeric_columns_summary = NB07_SUMMARY_DF.select_dtypes(include="number").columns.tolist()
if numeric_columns_summary:
    NB07_SUMMARY_DF[numeric_columns_summary] = NB07_SUMMARY_DF[numeric_columns_summary].round(4)

NB07_SUMMARY_DF = NB07_SUMMARY_DF.rename(columns={
    _column: prettify_column_name(_column) for _column in NB07_SUMMARY_DF.columns
})

# =============================================
# SAVE AND DISPLAY THE NB07 FINAL SUMMARY TABLE
# =============================================

NB07_SUMMARY_DF.to_csv(NB07_SUMMARY_CSV, index=False)

display(Markdown("### NB07 FINAL SUMMARY TABLE"))
display(NB07_SUMMARY_DF)

print("NB07 SUMMARY SAVED TO:", NB07_SUMMARY_CSV)
print("CODE_BLOCK_D COMPLETE")
```

```python
# ============
# CODE_BLOCK_F
# ============

# ===============================================
# MARKDOWN CELL TEMPLATE - AZURE OPENAI NARRATIVE
# ===============================================

MARKDOWN_TEMPLATE_F = """## NB07 - Azure OpenAI Narrative

This block is an optional presentation extension that generates plain-English result summaries.
"""

display(Markdown(MARKDOWN_TEMPLATE_F))

# ===============================================================================================
# THIS BLOCK IS AN OPTIONAL PRESENTATION EXTENSION NOTEBOOK SUCCESS DOES NOT DEPEND ON THIS BLOCK
# ===============================================================================================

print("OPTIONAL AZURE OPENAI BLOCK - NOTEBOOK SUCCESS DOES NOT DEPEND ON THIS BLOCK.")

# ================================================================
# WRAP THE ENTIRE OPTIONAL AZURE OPENAI FLOW IN A TRY EXCEPT BLOCK
# ================================================================

try:

    # ========================================================
    # LOAD THE REQUIRED ENVIRONMENT VARIABLES FOR AZURE OPENAI
    # ========================================================

    azure_openai_key = os.environ.get("AZURE_OPENAI_KEY")
    azure_openai_endpoint = os.environ.get("AZURE_OPENAI_ENDPOINT")
    azure_openai_deployment = os.environ.get("AZURE_OPENAI_DEPLOYMENT") or os.environ.get("AZURE_OPENAI_CHAT_DEPLOYMENT")

    if not azure_openai_key or not azure_openai_endpoint or not azure_openai_deployment:
        print("AZURE OPENAI SKIPPED - REQUIRED ENVIRONMENT VARIABLES ARE NOT SET.")
    else:

        # ==========================================================
        # IMPORT THE AZURE OPENAI CLIENT ONLY WHEN CREDENTIALS EXIST
        # ==========================================================

        from openai import AzureOpenAI

        # ======================================================
        # LOAD THE SUMMARY AND THE OPTIONAL SUPPORTING ARTIFACTS
        # ======================================================

        if not NB07_SUMMARY_CSV.exists():
            raise FileNotFoundError(f"MISSING REQUIRED SUMMARY FILE: {NB07_SUMMARY_CSV}")

        summary_df = pd.read_csv(NB07_SUMMARY_CSV)
        tier1_entropy_df = pd.read_csv(TABLES_DIR / "notebook04_entropy_comparison.csv") if (TABLES_DIR / "notebook04_entropy_comparison.csv").exists() else pd.DataFrame()
        tier2_entropy_df = pd.read_csv(TABLES_DIR / "notebook06_factor_exposure_entropy.csv") if (TABLES_DIR / "notebook06_factor_exposure_entropy.csv").exists() else pd.DataFrame()
        h3_df = pd.read_csv(TABLES_DIR / "notebook03_h3_matched_rmse_mae.csv") if (TABLES_DIR / "notebook03_h3_matched_rmse_mae.csv").exists() else pd.DataFrame()

        # ======================================================
        # BUILD THE AZURE OPENAI CLIENT AND THE NARRATIVE PROMPT
        # ======================================================

        client = AzureOpenAI(
            api_key=azure_openai_key,
            azure_endpoint=azure_openai_endpoint,
            api_version="2024-02-01",
        )

        prompt_text = f"""You are writing a concise plain-English capstone summary.

Use the following saved artifacts only.

NB07 FINAL SUMMARY
{summary_df.to_string(index=False)}

H3 MATCHED WINDOW RESULTS
{h3_df.to_string(index=False) if not h3_df.empty else "H3 ARTIFACT NOT AVAILABLE"}

TIER 1 ENTROPY RESULTS
{tier1_entropy_df.to_string(index=False) if not tier1_entropy_df.empty else "TIER 1 ENTROPY ARTIFACT NOT AVAILABLE"}

TIER 2 ENTROPY RESULTS
{tier2_entropy_df.to_string(index=False) if not tier2_entropy_df.empty else "TIER 2 ENTROPY ARTIFACT NOT AVAILABLE"}

Return markdown with four short sections:
1. H1 forecast accuracy
2. H2 portfolio risk control
3. H3 sentiment contribution
4. Entropy diagnostics
"""

        # ==============================================
        # CALL THE AZURE OPENAI CHAT COMPLETION ENDPOINT
        # ==============================================

        response = client.chat.completions.create(
            model=azure_openai_deployment,
            messages=[
                {"role": "system", "content": "You produce concise markdown summaries for a capstone defense notebook."},
                {"role": "user", "content": prompt_text},
            ],
            temperature=0.2,
            max_tokens=700,
        )

        narrative_text = response.choices[0].message.content.strip()

        # =====================================================
        # DISPLAY THE OPTIONAL NARRATIVE INLINE IN THE NOTEBOOK
        # =====================================================

        display(Markdown("### OPTIONAL AZURE OPENAI NARRATIVE"))
        display(Markdown(narrative_text))
        print("AZURE OPENAI NARRATIVE GENERATED SUCCESSFULLY.")

except Exception as azure_openai_error:
    print("AZURE OPENAI BLOCK SKIPPED - OPTIONAL EXTENSION FAILED SAFELY.")
    print("DETAILS:", azure_openai_error)

print("CODE_BLOCK_F COMPLETE")
```

```python
# ============
# CODE_BLOCK_G
# ============

# ===============================================
# MARKDOWN CELL TEMPLATE - NB07 ARTIFACT MANIFEST
# ===============================================

MARKDOWN_TEMPLATE_G = """## NB07 - Artifact Manifest

This final block builds the Notebook 07 manifest and enforces the Artifact-Truth Rule.
"""

display(Markdown(MARKDOWN_TEMPLATE_G))

# ============================================================
# DEFINE THE REPORT SECTION MAP FOR INPUT AND OUTPUT ARTIFACTS
# ============================================================

REPORT_SECTION_MAP_G = {
    "notebook04_causal_vs_baseline_rmse_mae.csv": "Table 16",
    "notebook04_asset_class_comparison.csv": "Table 17",
    "notebook04_causal_feature_importance.csv": "Table 18",
    "notebook04_entropy_comparison.csv": "Table 19",
    "notebook04_dag_gating_summary.csv": "Table 20",
    "notebook05_portfolio_performance.csv": "Table 21",
    "notebook05_risk_targeting_stability.csv": "Table 22",
    "notebook05_turnover_summary.csv": "Table 23",
    "notebook06_ablation_table.csv": "Table 24",
    "notebook06_dm_test_results.csv": "Table 25",
    "notebook07_composite_fig19_pipeline.png": "Figure 19",
    "notebook07_composite_fig20_dag.png": "Figure 20",
    "notebook07_composite_fig21_equity.png": "Figure 21",
    "notebook07_composite_fig22_drawdown.png": "Figure 22",
    "notebook07_composite_fig23_realized_vs_target.png": "Figure 23",
    "notebook07_composite_fig24_factor_entropy.png": "Figure 24",
    "notebook07_composite_fig25_ablation.png": "Figure 25",
    "notebook07_composite_fig26_regime_portfolio.png": "Figure 26",
    "notebook07_composite_fig27_stress.png": "Figure 27",
}

# ===============================================================
# DEFINE THE NB07 OUTPUT REGISTRY THAT MUST EXIST BY NOTEBOOK END
# ===============================================================

NB07_OUTPUT_REGISTRY_G = [
    ("NB07", "OUTPUT", FIGURES_DIR / "notebook07_composite_fig19_pipeline.png"),
    ("NB07", "OUTPUT", FIGURES_DIR / "notebook07_composite_fig20_dag.png"),
    ("NB07", "OUTPUT", FIGURES_DIR / "notebook07_composite_fig21_equity.png"),
    ("NB07", "OUTPUT", FIGURES_DIR / "notebook07_composite_fig22_drawdown.png"),
    ("NB07", "OUTPUT", FIGURES_DIR / "notebook07_composite_fig23_realized_vs_target.png"),
    ("NB07", "OUTPUT", FIGURES_DIR / "notebook07_composite_fig24_factor_entropy.png"),
    ("NB07", "OUTPUT", FIGURES_DIR / "notebook07_composite_fig25_ablation.png"),
    ("NB07", "OUTPUT", FIGURES_DIR / "notebook07_composite_fig26_regime_portfolio.png"),
    ("NB07", "OUTPUT", FIGURES_DIR / "notebook07_composite_fig27_stress.png"),
    ("NB07", "OUTPUT", NB07_SUMMARY_CSV),
    ("NB07", "OUTPUT", STREAMLIT_APP),
]

# ============================================================
# DEFINE A SMALL HELPER THAT RESOLVES INPUT FILENAMES TO PATHS
# ============================================================

def resolve_input_path(filename):

    # =========================================================
    # USE THE FILE EXTENSION TO INFER THE MOST LIKELY DIRECTORY
    # =========================================================

    filename_lower = str(filename).lower()
    if filename_lower.endswith(".csv"):
        return TABLES_DIR / filename
    if filename_lower.endswith(".png"):
        return FIGURES_DIR / filename
    if filename_lower.endswith(".parquet"):
        return DATA_DIR / filename
    return REPO_ROOT / filename

# ===============================================================
# BUILD THE INPUT MANIFEST ROWS FROM THE EXISTING AUDIT DATAFRAME
# ===============================================================

input_rows_g = []

for _row in AUDIT_DF.itertuples(index=False):
    source_notebook = str(_row.SOURCE).split(" ")[0]
    filename = str(_row.FILENAME)
    path = resolve_input_path(filename)
    exists_flag = bool(_row.IS_PRESENT)
    file_size_kb = round(path.stat().st_size / 1024, 1) if exists_flag and path.exists() else pd.NA
    input_rows_g.append({
        "filename": filename,
        "source_notebook": source_notebook,
        "artifact_role": "INPUT",
        "used_in_report_section": REPORT_SECTION_MAP_G.get(filename, "N/A"),
        "exists_flag": exists_flag,
        "file_size_kb": file_size_kb,
    })

# ============================================================
# BUILD THE OUTPUT MANIFEST ROWS FROM THE NB07 OUTPUT REGISTRY
# ============================================================

output_rows_g = []

for _source_notebook, _artifact_role, _path in NB07_OUTPUT_REGISTRY_G:
    exists_flag = _path.exists()
    file_size_kb = round(_path.stat().st_size / 1024, 1) if exists_flag else pd.NA
    output_rows_g.append({
        "filename": _path.name,
        "source_notebook": _source_notebook,
        "artifact_role": _artifact_role,
        "used_in_report_section": REPORT_SECTION_MAP_G.get(_path.name, "N/A"),
        "exists_flag": exists_flag,
        "file_size_kb": file_size_kb,
    })

# ==============================================================
# COMBINE INPUT AND OUTPUT ROWS INTO THE NB07 MANIFEST DATAFRAME
# ==============================================================

NB07_MANIFEST_DF = pd.DataFrame(input_rows_g + output_rows_g)

# =========================================================
# SAVE THE MANIFEST ONCE SO THE FILE ITSELF CAN BE RECORDED
# =========================================================

NB07_MANIFEST_DF.to_csv(NB07_MANIFEST_CSV, index=False)

# ==========================================================
# APPEND THE MANIFEST FILE ITSELF AS A FINAL NB07 OUTPUT ROW
# ==========================================================

manifest_file_size_kb = round(NB07_MANIFEST_CSV.stat().st_size / 1024, 1) if NB07_MANIFEST_CSV.exists() else pd.NA
manifest_self_row = pd.DataFrame([{
    "filename": NB07_MANIFEST_CSV.name,
    "source_notebook": "NB07",
    "artifact_role": "OUTPUT",
    "used_in_report_section": "N/A",
    "exists_flag": NB07_MANIFEST_CSV.exists(),
    "file_size_kb": manifest_file_size_kb,
}])

NB07_MANIFEST_DF = pd.concat([NB07_MANIFEST_DF, manifest_self_row], ignore_index=True)
NB07_MANIFEST_DF.to_csv(NB07_MANIFEST_CSV, index=False)

# ===========================================================
# DISPLAY THE FINAL MANIFEST AND PRINT THE COMPLETION MESSAGE
# ===========================================================

display(NB07_MANIFEST_DF)
print("NB07 MANIFEST SAVED TO:", NB07_MANIFEST_CSV)
print("✅ NB07 COMPLETE — ALL OUTPUTS CONFIRMED.")
```
